### Feature engineering

***Task*** - Create financial ratios

Calculate the financial ratios:

**Profitability Ratios**
* Net Profit Margin = Net Income / Revenue
    * The Net Profit Margin measures a company's profitability by comparing its net income to its revenue. A higher net profit margin indicates that a company is generating more profit from its sales revenue.
* Gross Profit Margin = Gross Income / Revenue
    * The Gross Profit Margin measures a company's profitability by comparing its gross income to its revenue. A higher gross profit margin indicates that a company is generating more profit from its sales revenue before accounting for operating expenses.
* Return on equity = Net Income / Equity
    * The Return on Equity measures a company's profitability by comparing its net income to its equity. A higher return on equity indicates that a company is generating more profit from its shareholders' investment.
* Return on asset = Net Income / Total Assets
    * The Return on Asset measures a company's profitability by comparing its net income to its total assets. A higher return on asset indicates that a company is generating more profit from its total assets.

**Liquidity Ratios**
* Current Ratio = Current Assets / Current Liabilities
    * The Current Ratio measures a company's liquidity by comparing its current assets to its current liabilities. A higher current ratio indicates that a company has sufficient liquid assets to meet its short-term obligations.
* Quick Ratio = (Current Assets - Inventory) / Current Liabilities
    * The Quick Ratio measures a company's liquidity by comparing its quick assets (current assets minus inventory) to its current liabilities. A higher quick ratio indicates that a company has sufficient liquid assets to meet its short-term obligations.

**Solvency/Leverage Ratios**
* Debt-to-Equity (D/E) Ratio = Total Liabilities / Equity
    * The Debt-to-Equity Ratio measures a company's leverage by comparing its total liabilities to its equity. A higher debt-to-equity ratio indicates that a company is more heavily leveraged.
* Debt-to-Asset Ratio = Total Liabilities / Total Assets
    * The Debt-to-Asset Ratio measures a company's leverage by comparing its total liabilities to its total assets. A higher debt-to-asset ratio indicates that a company is more heavily leveraged.

**Efficiency ratio**
* Asset Turnover Ratio = Net Sales / Average Total Assets
    * The Asset Turnover Ratio measures a company's efficiency by comparing its net sales to its average total assets. A higher asset turnover ratio indicates that a company is generating more sales revenue from its total assets.
* Inventory Turnover = Cost of Goods Sold / Average Inventory
    * The Inventory Turnover measures a company's efficiency by comparing its cost of goods sold to its average inventory. A higher inventory turnover indicates that a company is selling and replacing its inventory quickly.
* Receivables Turnover = Net Credit Sales / Average Accounts Receivable
    * The Receivables Turnover measures a company's efficiency by comparing its net credit sales to its average accounts receivable. A higher receivables turnover indicates that a company is collecting its accounts receivable quickly.
